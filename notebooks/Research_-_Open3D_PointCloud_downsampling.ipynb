# PointCloud downsampling with Open3D library 

```python
import numpy as np
import open3d as o3d

from os import environ

from plant3dvision.visu import plotly_pointcloud
from plant3dvision.utils import locate_task_filesets
from plantdb.commons.fsdb import FSDB
from plantdb.commons.io import read_point_cloud
```

```python
db = FSDB(environ.get('ROMI_DB', "/data/ROMI/Romi_Alexis/analyse_jo"))
db.connect(unsafe=True)
```

```python
scan = db.get_scan('Col-0_E1_1', create=False)
```

```python
fileset_names = locate_task_filesets(scan, ["PointCloud"])
pcd_fs = scan.get_fileset(fileset_names['PointCloud'])
```

```python
pcd = read_point_cloud(pcd_fs.get_file('PointCloud'))
```

```python
db.disconnect()
```

```python
n_pts = {0: len(pcd.points)}
for vxs in np.arange(0.01, 2., 0.01):
    low_res_pcd = pcd.voxel_down_sample(vxs)
    n_pts[vxs] = len(low_res_pcd.points)
```

```python
import matplotlib.pyplot as plt

plt.scatter(list(n_pts.keys()), list(n_pts.values()))
plt.show()
```

```python
vxs=1.5
fig = plotly_pointcloud(pcd.voxel_down_sample(vxs), n_pts='all')
```
