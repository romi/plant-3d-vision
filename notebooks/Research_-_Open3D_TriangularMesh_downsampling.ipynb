# TriangularMesh downsampling with Open3D library 

```python
import numpy as np
import open3d as o3d

from os import environ

from plant3dvision.utils import locate_task_filesets
from plantdb.commons.fsdb import FSDB
from plantdb.commons.io import read_triangle_mesh
```

```python
db = FSDB(environ.get('ROMI_DB', "/data/ROMI/Romi_Alexis/analyse_jo"))
db.connect(unsafe=True)
```

```python
scan = db.get_scan('Col-0_E1_1', create=False)
```

```python
fileset_names = locate_task_filesets(scan, ["TriangleMesh"])
mesh_fs = scan.get_fileset(fileset_names['TriangleMesh'])
```

```python
mesh = read_triangle_mesh(mesh_fs.get_file('TriangleMesh'))
```

```python
db.disconnect()
```

```python
n_pts = {0: len(mesh.vertices)}
for n_iter in np.arange(1, 5):
    low_res_mesh = mesh.filter_smooth_taubin(n_iter, 0.5, -0.53)
    n_pts[n_iter] = len(low_res_mesh.vertices)
```

```python
import matplotlib.pyplot as plt

plt.scatter(list(n_pts.keys()), list(n_pts.values()))
plt.show()
```
